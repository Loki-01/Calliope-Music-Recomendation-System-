We want to know which decade each song belongs to.
For example:
year 1984  ->  decade "1980s"
year 2003  ->  decade "2000s"
year 2021  ->  decade "2020s"

```python
data["decade"] = data["year"].apply(lambda x: f"{(x // 10) * 10}s")
```
